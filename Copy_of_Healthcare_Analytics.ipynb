```python

import pandas as pd
import os

csv_files = [
    f for f in os.listdir("hmis_data")
    if f.endswith(".csv") and f != "metadata.csv"
]

file_path = os.path.join("hmis_data", csv_files[0])

df = pd.read_csv(file_path, low_memory=False)

print("Dataset Shape:", df.shape)
display(df.head())
```

```python

print("Dataset Information:")
df.info()

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Records:")
print(df.duplicated().sum())

print("\nDistricts:", df["district_as_per_lgd"].nunique())

print("\nFinancial Years:")
print(sorted(df["fiscal_year"].unique()))
```

```python

# Step 2: Missing Data Analysis

missing_summary = pd.DataFrame({
    "Missing_Count": df.isnull().sum(),
    "Missing_Percentage": (df.isnull().mean() * 100).round(2)
})

print("MISSING DATA REPORT")
display(missing_summary)

print("\nDataset Statistical Summary:")
display(df.describe(include="all").T)
```

```python

# Step 3: Explore Healthcare Indicators

print("Total Unique Indicators:")
print(df["indicator"].nunique())

print("\nHealthcare Indicator Categories:")
display(df["indicator_head"].value_counts())

print("\nDistrict-wise Record Count:")
display(df["district_as_per_lgd"].value_counts())
```

```python

# Step 4: Identify Important Healthcare Indicators

indicator_summary = (
    df.groupby(["indicator_head", "indicator"])
    .size()
    .reset_index(name="record_count")
    .sort_values("record_count", ascending=False)
)

print("Total Indicator Combinations:",
      len(indicator_summary))

print("\nAll OPD / IPD Related Indicators:")

keywords = r"OPD|IPD|OUTPATIENT|INPATIENT|OUT PATIENT|IN PATIENT"

opd_ipd = indicator_summary[
    indicator_summary["indicator"].str.contains(
        keywords, case=False, na=False
    )
]

display(opd_ipd)

print("\nTop 30 Healthcare Indicators:")
display(indicator_summary.head(30))
```

```python

# Step 5: Find OPD and IPD Attendance Indicators

attendance_keywords = (
    r"outpatient attendance|inpatient attendance|"
    r"patients admitted|inpatient.*adult|"
    r"ipd attendance|opd attendance"
)

attendance_data = indicator_summary[
    indicator_summary["indicator"].str.contains(
        attendance_keywords,
        case=False,
        na=False,
        regex=True
    )
]

display(attendance_data)

print("Matching indicators:", len(attendance_data))
```

```python

# Step 6: Inspect actual OPD attendance records

opd_rows = df[
    df["indicator"].str.contains(
        "Outpatient attendance",
        case=False,
        na=False
    )
]

display(
    opd_rows[
        ["fiscal_year", "district_as_per_lgd",
         "indicator", "facility_category",
         "category", "value", "unit"]
    ].head(15)
)
```

```python

# Step 7: Explore IPD Attendance and Admissions

ipd_keywords = (
    r"Total number of patients admitted|"
    r"Inpatient.*Adults|"
    r"Inpatient attendance|"
    r"IPD attendance"
)

ipd_rows = df[
    df["indicator"].str.contains(
        ipd_keywords,
        case=False,
        na=False,
        regex=True
    )
]

display(
    ipd_rows[
        [
            "fiscal_year",
            "district_as_per_lgd",
            "indicator",
            "facility_category",
            "category",
            "value",
            "unit"
        ]
    ].head(25)
)

print("Total IPD matching records:", len(ipd_rows))
```

```python

# Step 7: Explore IPD Attendance and Admissions

ipd_keywords = (
    r"Total number of patients admitted|"
    r"Inpatient.*Adults|"
    r"Inpatient attendance|"
    r"IPD attendance"
)

ipd_rows = df[
    df["indicator"].str.contains(
        ipd_keywords,
        case=False,
        na=False,
        regex=True
    )
]

display(
    ipd_rows[
        [
            "fiscal_year",
            "district_as_per_lgd",
            "indicator",
            "facility_category",
            "category",
            "value",
            "unit"
        ]
    ].head(25)
)

print("Total IPD matching records:", len(ipd_rows))
```

```python

# Restore OPD and IPD variables

opd_rows = df[
    df["indicator"].str.contains(
        "Outpatient attendance",
        case=False,
        na=False
    )
].copy()

ipd_data = df[
    df["indicator"].str.contains(
        r"inpatient|patients admitted|ipd",
        case=False,
        na=False
    )
].copy()

print("OPD Records:", len(opd_rows))
print("IPD Records:", len(ipd_data))

print("\nOPD Facility Categories:")
display(opd_rows["facility_category"].value_counts())

print("\nIPD Facility Categories:")
display(ipd_data["facility_category"].value_counts())

print("\nAvailable Financial Years:")
display(sorted(df["fiscal_year"].dropna().unique()))
```
